What does this map mean?

This is a snapshot of sea-surface temperature (SST) on 15 July 2024 over our selected Indian Ocean region.

The colors represent temperature:

🔵 Blue/purple → colder water, roughly 10–18°C
🟢 Green → ~18–22°C
🟡 Yellow → ~23–26°C
🟠 Orange → ~27–30°C
🔴 Red → ~30–35°C

So, for example, the orange/red areas around the tropical Indian Ocean indicate warmer surface water, while the blue/green areas toward the southern part are considerably cooler.

And notice something important:

This is only surface temperature. It tells us what the ocean surface looks like — not what the temperature is 500 m or 1000 m below.**bold text**

```python
import xarray as xr
import matplotlib.pyplot as plt

sst2024 = xr.open_dataset('/content/sst2024_indian.nc')['sst']

dates = ['2024-01-15', '2024-04-15', '2024-07-15', '2024-10-15']
titles = ['January 2024', 'April 2024', 'July 2024', 'October 2024']

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

for ax, date, title in zip(axes.flat, dates, titles):
    sst2024.sel(time=date).plot(
        ax=ax,
        cmap='turbo',
        vmin=15,
        vmax=35,
        add_colorbar=False
    )
    ax.set_title(f'NOAA OISST v2.1 — {title}')
    ax.set_xlabel('Longitude (°E)')
    ax.set_ylabel('Latitude (°)')

plt.tight_layout()

plt.savefig(
    '/content/SST_2024_seasonal_maps.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

**How does the average Indian Ocean surface temperature change throughout 2024 and 2025?**

```python
import xarray as xr
import matplotlib.pyplot as plt

sst2024 = xr.open_dataset('/content/sst2024_indian.nc')['sst']
sst2025 = xr.open_dataset('/content/sst2025_indian.nc')['sst']

mean2024 = sst2024.mean(dim=['lat', 'lon'], skipna=True)
mean2025 = sst2025.mean(dim=['lat', 'lon'], skipna=True)

plt.figure(figsize=(15, 6))

plt.plot(
    mean2024.time,
    mean2024,
    label='2024'
)

plt.plot(
    mean2025.time,
    mean2025,
    label='2025'
)

plt.title('Average Sea Surface Temperature — Indian Ocean')
plt.xlabel('Date')
plt.ylabel('Average SST (°C)')
plt.legend()
plt.grid(True, alpha=0.3)

plt.savefig(
    '/content/SST_2024_2025_timeseries.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
import xarray as xr
import matplotlib.pyplot as plt

sst2024 = xr.open_dataset('/content/sst2024_indian.nc')['sst']
sst2025 = xr.open_dataset('/content/sst2025_indian.nc')['sst']

monthly2024 = sst2024.resample(time='1MS').mean().mean(dim=['lat', 'lon'], skipna=True)
monthly2025 = sst2025.resample(time='1MS').mean().mean(dim=['lat', 'lon'], skipna=True)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly2024.time,
    monthly2024,
    marker='o',
    label='2024'
)

plt.plot(
    monthly2025.time,
    monthly2025,
    marker='o',
    label='2025'
)

plt.title('Monthly Average Sea Surface Temperature — Indian Ocean')
plt.xlabel('Month')
plt.ylabel('Average SST (°C)')
plt.legend()
plt.grid(True, alpha=0.3)

plt.savefig(
    '/content/SST_monthly_2024_2025.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

COMDINE SST


```python
import xarray as xr

sst2024 = xr.open_dataset("/content/sst2024_indian.nc")
sst2025 = xr.open_dataset("/content/sst2025_indian.nc")

sst_combined = xr.concat([sst2024, sst2025], dim="time").sortby("time")

print(sst_combined)
```

```python
sst_data = sst_combined["sst"]

print("Total values:", sst_data.size)
print("NaN values:", sst_data.isnull().sum().compute().item())
print("NaN percentage:", (sst_data.isnull().mean().compute().item() * 100))
```

```python
print("Minimum SST:", sst_data.min().compute().item())
print("Maximum SST:", sst_data.max().compute().item())
print("Mean SST:", sst_data.mean().compute().item())
```

```python
nan_by_location = sst_data.isnull().mean(dim="time").compute()

print("Minimum NaN % at a grid cell:", float(nan_by_location.min()) * 100)
print("Maximum NaN % at a grid cell:", float(nan_by_location.max()) * 100)
```

```python
permanent_nan = (nan_by_location == 1).sum().item()
total_grid_cells = nan_by_location.size

print("Total grid cells:", total_grid_cells)
print("100% NaN grid cells:", permanent_nan)
print("Percentage of grid cells permanently NaN:", permanent_nan / total_grid_cells * 100)
```

```python
non_permanent = nan_by_location.where(nan_by_location < 1, drop=True)

print("Grid cells with at least some valid SST:", non_permanent.size)
print("Maximum NaN % among these cells:", float(non_permanent.max()) * 100)
print("Mean NaN % among these cells:", float(non_permanent.mean()) * 100)
```

```python
zero_nan = (nan_by_location == 0).sum().item()
partial_nan = ((nan_by_location > 0) & (nan_by_location < 1)).sum().item()
full_nan = (nan_by_location == 1).sum().item()

print("Grid cells with 0% NaN:", zero_nan)
print("Grid cells with occasional NaN:", partial_nan)
print("Grid cells with 100% NaN:", full_nan)
print("Total grid cells:", zero_nan + partial_nan + full_nan)
```

```python
sst_save = sst_combined.copy()

sst_save["sst"].encoding.pop("_FillValue", None)
sst_save["sst"].encoding.pop("missing_value", None)

sst_save.to_netcdf("/content/SST_IndianOcean_2024_2025.nc")

print("SST dataset saved successfully!")
```

```python
import requests
import xarray as xr
import os
import time

os.makedirs("/content/SSS_2024_IndianOcean", exist_ok=True)

for day in range(1, 367):
    filename = f"SP_D2024{day:03d}_Map_SATSSS_data_1day.nc"
    url = f"https://www.star.nesdis.noaa.gov/data/socd1/coastwatch/products/smap/nc/{filename}"
    raw_file = f"/content/{filename}"
    output_file = f"/content/SSS_2024_IndianOcean/{filename}"

    if os.path.exists(output_file):
        print(f"Already done: {day}/366")
        continue

    try:
        r = requests.get(url, timeout=180)

        if r.status_code != 200:
            print(f"❌ Day {day}: HTTP {r.status_code}")
            continue

        with open(raw_file, "wb") as f:
            f.write(r.content)

        ds = xr.open_dataset(raw_file)

        sss_indian = ds["sss"].sel(
            lat=slice(-40, 30),
            lon=slice(20, 120)
        )

        sss_indian.to_netcdf(output_file)

        ds.close()
        os.remove(raw_file)

        print(f"✅ Day {day}/366 completed")

    except Exception as e:
        print(f"❌ Day {day} failed: {e}")

    time.sleep(0.5)

print("🎉 2024 SSS collection completed!")
```

```python
import requests
import xarray as xr
import os
import time

os.makedirs("/content/SSS_2024_IndianOcean", exist_ok=True)

for day in range(1, 367):
    filename = f"SP_D2024{day:03d}_Map_SATSSS_data_1day.nc"
    url = f"https://www.star.nesdis.noaa.gov/data/socd1/coastwatch/products/smap/nc/{filename}"
    raw_file = f"/content/{filename}"
    output_file = f"/content/SSS_2024_IndianOcean/{filename}"

    if os.path.exists(output_file):
        print(f"Already done: {day}/366")
        continue

    try:
        print(f"\nDownloading Day {day}/366...")

        r = requests.get(url, timeout=180)

        if r.status_code != 200:
            print(f"❌ Day {day}: HTTP {r.status_code}")
            continue

        with open(raw_file, "wb") as f:
            f.write(r.content)

        try:
            ds = xr.open_dataset(raw_file)

            sss_indian = ds["sss"].sel(
                lat=slice(-40, 30),
                lon=slice(20, 120)
            ).load()

            ds.close()

            sss_indian.to_netcdf(output_file)

            os.remove(raw_file)

            print(f"✅ Day {day}/366 completed")

        except Exception as e:
            print(f"⚠️ Day {day} downloaded but could not be read: {e}")

            try:
                ds.close()
            except:
                pass

            if os.path.exists(raw_file):
                os.remove(raw_file)

    except Exception as e:
        print(f"❌ Day {day} failed: {e}")

    time.sleep(1)
```

```python
import xarray as xr
import glob
import os

for year in [2024, 2025]:
    folder = f"/content/SSS_{year}_IndianOcean"
    files = sorted(glob.glob(folder + "/*.nc"))

    print(f"\n===== {year} =====")
    print("Files:", len(files))

    bad_files = []

    for f in files:
        try:
            ds = xr.open_dataset(f)

            lat = ds["lat"].values
            lon = ds["lon"].values

            duplicate_lat = len(lat) != len(set(lat))
            duplicate_lon = len(lon) != len(set(lon))

            if duplicate_lat or duplicate_lon:
                bad_files.append((
                    os.path.basename(f),
                    duplicate_lat,
                    duplicate_lon,
                    len(lat),
                    len(lon)
                ))

            ds.close()

        except Exception as e:
            bad_files.append((
                os.path.basename(f),
                "ERROR",
                str(e)
            ))

    print("Problematic files:", len(bad_files))

    for item in bad_files[:20]:
        print(item)
```

```python
import xarray as xr
import glob
import numpy as np

files_2024 = sorted(
    glob.glob("/content/SSS_2024_IndianOcean/*.nc")
)

sss2024_parts = []

global_count = 0
cropped_count = 0

for f in files_2024:
    ds = xr.open_dataset(f)

    # Full global file
    if len(ds.lat) == 720 and len(ds.lon) == 1440:
        ds = ds.sel(
            lat=slice(-40, 30),
            lon=slice(20, 120)
        )
        global_count += 1

    # Already cropped file
    elif len(ds.lat) == 280 and len(ds.lon) == 400:
        cropped_count += 1

    else:
        print("Unexpected grid:", f, len(ds.lat), len(ds.lon))
        ds.close()
        continue

    # Keep only SSS
    ds = ds[["sss"]]

    sss2024_parts.append(ds)

print("===== 2024 NORMALIZATION =====")
print("Total files:", len(sss2024_parts))
print("Global files cropped:", global_count)
print("Already cropped:", cropped_count)

print("\nFirst normalized grid:")
print(sss2024_parts[0])

print("\nGrid size of first file:")
print("Latitude:", len(sss2024_parts[0].lat))
print("Longitude:", len(sss2024_parts[0].lon))
```

1. What the map is showing

This is Sea Surface Salinity (SSS) over our selected Indian Ocean region:

20°E–120°E and 40°S–30°N

The color bar represents salinity:

🟣 Purple → lower salinity
🟢 Green → medium salinity
🟡 Yellow → higher salinity
Scale shown is roughly 30–38

So each pixel represents the estimated surface salinity at that location.

2. The annual average map

The top-left map is the most important one for understanding the overall 2024 pattern.

It is much smoother than the daily maps because we're averaging many observations throughout the year.

From it, we can see:

Large parts of the Indian Ocean are around the green/teal range, roughly mid-30s.
There are lower-salinity regions toward parts of the northern/eastern Indian Ocean, visible as darker blue/purple.
Some regions toward the western/southern Indian Ocean show relatively higher salinity, moving toward yellow.
The spatial pattern isn't uniform — salinity changes substantially depending on location.

So the annual map basically answers:

"What does the typical surface salinity distribution look like across our Indian Ocean region during 2024?"

3. The January / April / July / October maps

These are different because they're individual daily observations, not annual averages.

And this is actually important for our project.

You can see that the pattern changes between:

January → April → July → October

For example, some regions become greener/yellower while others become darker.

That means:

Surface salinity is not constant throughout the year.

It has temporal variability.

That's useful because our eventual ML model shouldn't just learn:

Location → subsurface temperature

It should eventually learn something more like:

Surface ocean state + location + time
                    ↓
       subsurface temperature
4. What are those huge white gaps? ⚠️

This is VERY important.

Those white regions in the January/April/July/October maps should not be interpreted as "salinity = 0".

They represent missing/unavailable observations in those particular daily satellite products.

That's why we did the missing-value analysis earlier.

The annual average looks much more complete because we're averaging across many available observations.

So:

Daily satellite map
       ↓
Some locations unavailable
       ↓
Lots of NaNs

Many daily maps
       ↓
Average available observations
       ↓
Much more complete annual map

This is actually a nice thing to explain during your presentation.

5. The biggest thing I understand from these maps

Your maps demonstrate two dimensions of variability:

Spatial variability

Different parts of the Indian Ocean have different surface salinity.

Temporal variability

The surface salinity pattern changes across the year.

And that's exactly why SSS can potentially be useful as one of the surface input features for your reconstruction model.

```python
import json
import os
import numpy as np
import pandas as pd

folder = "/content/Argo_IndianOcean"
files = sorted([f for f in os.listdir(folder) if f.endswith(".json")])

target_depths = [50, 100, 200, 500, 1000, 1500, 2000]

rows = []

for file in files:
    with open(os.path.join(folder, file), "r") as f:
        profiles = json.load(f)

    for profile in profiles:
        data = profile.get("data")

        if not data or len(data) < 2:
            continue

        pressure = np.array(data[0], dtype=float)
        temperature = np.array(
            [np.nan if x is None else x for x in data[1]],
            dtype=float
        )

        valid = np.isfinite(pressure) & np.isfinite(temperature)

        if valid.sum() < 2:
            continue

        p = pressure[valid]
        t = temperature[valid]

        order = np.argsort(p)
        p = p[order]
        t = t[order]

        # Remove duplicate pressure values
        p_unique, idx = np.unique(p, return_index=True)
        t_unique = t[idx]

        values = {}

        for depth in target_depths:
            if p_unique.min() <= depth <= p_unique.max():
                values[f"T_{depth}"] = np.interp(
                    depth, p_unique, t_unique
                )
            else:
                values[f"T_{depth}"] = np.nan

        coordinates = profile.get("geolocation", {}).get("coordinates", [np.nan, np.nan])

        rows.append({
            "profile_id": profile.get("_id"),
            "longitude": coordinates[0],
            "latitude": coordinates[1],
            "timestamp": profile.get("timestamp"),
            **values
        })

argo_df = pd.DataFrame(rows)

print("Dataset shape:", argo_df.shape)
print("\nColumns:")
print(argo_df.columns.tolist())

print("\nFirst 5 rows:")
display(argo_df.head())

print("\nMissing values:")
print(argo_df.isna().sum())
```

🌊 Why are we using dbar?

An Argo float measures pressure as it goes deeper into the ocean.

A very useful oceanographic approximation is:

1 dbar ≈ 1 meter of seawater depth

So for our project:

Pressure	Approx. depth	Our variable
50 dbar	~50 m	T_50
100 dbar	~100 m	T_100
200 dbar	~200 m	T_200
500 dbar	~500 m	T_500
1000 dbar	~1000 m	T_1000
1500 dbar	~1500 m	T_1500
2000 dbar	~2000 m	T_2000

6.24–6.48°C

```python
import numpy as np
import pandas as pd

ml_data = combined.copy()

ml_data["date"] = pd.to_datetime(ml_data["date"])

day_of_year = ml_data["date"].dt.dayofyear

ml_data["day_sin"] = np.sin(2 * np.pi * day_of_year / 365.25)
ml_data["day_cos"] = np.cos(2 * np.pi * day_of_year / 365.25)

print(ml_data[[
    "date",
    "day_sin",
    "day_cos"
]].head())

print("\nTime features added successfully.")
```

```python
train = ml_data[
    (ml_data["date"] >= "2024-01-01") &
    (ml_data["date"] < "2025-01-01")
].copy()

validation = ml_data[
    (ml_data["date"] >= "2025-01-01") &
    (ml_data["date"] < "2025-07-01")
].copy()

test = ml_data[
    (ml_data["date"] >= "2025-07-01") &
    (ml_data["date"] < "2026-01-01")
].copy()

print("TRAIN:", len(train))
print("VALIDATION:", len(validation))
print("TEST:", len(test))

print("\nDate ranges:")
print("Train:", train["date"].min(), "→", train["date"].max())
print("Validation:", validation["date"].min(), "→", validation["date"].max())
print("Test:", test["date"].min(), "→", test["date"].max())
```

```python
features = [
    "SST",
    "SSS_clean",
    "latitude",
    "longitude",
    "day_sin",
    "day_cos"
]

targets = [
    "T_50",
    "T_100",
    "T_200",
    "T_500",
    "T_1000",
    "T_1500"
]

train_v1 = train[features + targets].dropna().copy()
val_v1 = validation[features + targets].dropna().copy()
test_v1 = test[features + targets].dropna().copy()

print("Training samples:", len(train_v1))
print("Validation samples:", len(val_v1))
print("Test samples:", len(test_v1))
```

```python
from sklearn.preprocessing import StandardScaler

X_train = train_v1[features]
y_train = train_v1[targets]

X_val = val_v1[features]
y_val = val_v1[targets]

X_test = test_v1[features]
y_test = test_v1[targets]

X_scaler = StandardScaler()
y_scaler = StandardScaler()

X_train_scaled = X_scaler.fit_transform(X_train)
X_val_scaled = X_scaler.transform(X_val)
X_test_scaled = X_scaler.transform(X_test)

y_train_scaled = y_scaler.fit_transform(y_train)
y_val_scaled = y_scaler.transform(y_val)
y_test_scaled = y_scaler.transform(y_test)

print("X train shape:", X_train_scaled.shape)
print("Y train shape:", y_train_scaled.shape)

print("\nX feature means:")
print(X_train_scaled.mean(axis=0))

print("\nX feature standard deviations:")
print(X_train_scaled.std(axis=0))
```

```python
import torch
from torch.utils.data import TensorDataset, DataLoader

X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train_scaled, dtype=torch.float32)

X_val_tensor = torch.tensor(X_val_scaled, dtype=torch.float32)
y_val_tensor = torch.tensor(y_val_scaled, dtype=torch.float32)

X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test_scaled, dtype=torch.float32)

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
val_dataset = TensorDataset(X_val_tensor, y_val_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=64, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
import torch.nn as nn

class OceanMLP(nn.Module):
    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(6, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 6)
        )

    def forward(self, x):
        return self.network(x)

model = OceanMLP()

print(model)
```

```python
criterion = nn.MSELoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

print("Loss function:", criterion)
print("Optimizer:", optimizer)
```

```python
import copy

epochs = 100
best_val_loss = float("inf")
best_model_state = None

train_losses = []
val_losses = []

for epoch in range(epochs):
    model.train()
    train_loss = 0.0

    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()

        predictions = model(X_batch)
        loss = criterion(predictions, y_batch)

        loss.backward()
        optimizer.step()

        train_loss += loss.item()

    train_loss /= len(train_loader)

    model.eval()
    val_loss = 0.0

    with torch.no_grad():
        for X_batch, y_batch in val_loader:
            predictions = model(X_batch)
            loss = criterion(predictions, y_batch)
            val_loss += loss.item()

    val_loss /= len(val_loader)

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_model_state = copy.deepcopy(model.state_dict())

    if (epoch + 1) % 10 == 0:
        print(
            f"Epoch [{epoch+1}/{epochs}] "
            f"Train Loss: {train_loss:.4f} "
            f"Validation Loss: {val_loss:.4f}"
        )

model.load_state_dict(best_model_state)

print("\nTraining completed.")
print("Best validation loss:", best_val_loss)
```

```python
cnn_patches = []
cnn_indices = []

for group_num, ((year, date), indices) in enumerate(profiles_by_date.items(), 1):

    # Select correct SST dataset and SSS folder
    if year == 2024:
        sst_data = sst2024
        sss_folder = "/content/SSS_2024_IndianOcean"
    else:
        sst_data = sst2025
        sss_folder = "/content/SSS_2025_IndianOcean"

    # Find SSS file
    doy = date.dayofyear
    files = glob.glob(
        f"{sss_folder}/SP_D{year}{doy:03d}_Map_SATSSS_data_1day.nc"
    )

    if not files:
        continue

    # Load SSS ONCE for this date
    with xr.open_dataset(files[0]) as ds:
        sss_day = ds["sss"].squeeze().load()

    # Load SST ONCE for this date
    sst_day = sst_data.sel(
        time=date,
        method="nearest"
    )["sst"]

    # Process all profiles for this date
    for idx in indices:

        row = combined.loc[idx]
        lat = row["latitude"]
        lon = row["longitude"]

        # ---------- SST ----------
        lat_idx = np.abs(
            sst_day.lat.values - lat
        ).argmin()

        lon_idx = np.abs(
            sst_day.lon.values - lon
        ).argmin()

        if (
            lat_idx < 4 or
            lat_idx >= len(sst_day.lat) - 4 or
            lon_idx < 4 or
            lon_idx >= len(sst_day.lon) - 4
        ):
            continue

        sst_patch = sst_day.isel(
            lat=slice(lat_idx - 4, lat_idx + 5),
            lon=slice(lon_idx - 4, lon_idx + 5)
        ).values

        if sst_patch.shape != (9, 9):
            continue

        if np.isnan(sst_patch).any():
            continue

        # ---------- SSS ----------
        sss_lat_idx = np.abs(
            sss_day.lat.values - lat
        ).argmin()

        sss_lon_idx = np.abs(
            sss_day.lon.values - lon
        ).argmin()

        if (
            sss_lat_idx < 4 or
            sss_lat_idx >= len(sss_day.lat) - 4 or
            sss_lon_idx < 4 or
            sss_lon_idx >= len(sss_day.lon) - 4
        ):
            continue

        sss_patch = sss_day.isel(
            lat=slice(sss_lat_idx - 4, sss_lat_idx + 5),
            lon=slice(sss_lon_idx - 4, sss_lon_idx + 5)
        )

        if sss_patch.shape != (9, 9):
            continue

        missing_fraction = float(
            sss_patch.isnull().mean().values
        )

        if missing_fraction > 0.10:
            continue

        # Fill small gaps
        sss_patch = sss_patch.interpolate_na(
            dim="lat",
            method="nearest",
            fill_value="extrapolate"
        ).interpolate_na(
            dim="lon",
            method="nearest",
            fill_value="extrapolate"
        )

        sss_values = sss_patch.values

        if np.isnan(sss_values).any():
            continue

        # Save patch + original profile index
        cnn_patches.append(
            np.stack([
                sst_patch,
                sss_values
            ]).astype(np.float32)
        )

        cnn_indices.append(idx)

    # Progress
    if group_num % 50 == 0:
        print(
            f"Processed {group_num}/{len(profiles_by_date)} dates | "
            f"Valid patches: {len(cnn_patches)}"
        )

print("\nDONE!")

X_CNN = np.stack(cnn_patches).astype(np.float32)

print("Total profiles:", len(combined))
print("Valid CNN samples:", len(cnn_indices))
print("Rejected:", len(combined) - len(cnn_indices))
print("X_CNN shape:", X_CNN.shape)
print("Missing values:", np.isnan(X_CNN).sum())
```

```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

# Convert NumPy arrays to PyTorch tensors
X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
Y_train_tensor = torch.tensor(Y_train_scaled, dtype=torch.float32)

X_val_tensor = torch.tensor(X_val_scaled, dtype=torch.float32)
Y_val_tensor = torch.tensor(Y_val_scaled, dtype=torch.float32)

X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)
Y_test_tensor = torch.tensor(Y_test_scaled, dtype=torch.float32)

# Create datasets
train_dataset = TensorDataset(
    X_train_tensor,
    Y_train_tensor
)

val_dataset = TensorDataset(
    X_val_tensor,
    Y_val_tensor
)

test_dataset = TensorDataset(
    X_test_tensor,
    Y_test_tensor
)

# Create DataLoaders
train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=64,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
class OceanCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(2, 32, kernel_size=3, padding=1),
            nn.ReLU(),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),

            nn.MaxPool2d(kernel_size=2)
        )

        self.regressor = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 4 * 4, 128),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(128, 6)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.regressor(x)
        return x


model = OceanCNN()

print(model)

# Test one batch
sample_batch, sample_target = next(iter(train_loader))
sample_output = model(sample_batch)

print("\nInput shape:", sample_batch.shape)
print("Output shape:", sample_output.shape)
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = OceanCNN().to(device)

criterion = nn.MSELoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

best_val_loss = float("inf")
best_state = None

train_losses = []
val_losses = []

epochs = 100

for epoch in range(epochs):

    # ---------- Training ----------
    model.train()
    train_loss = 0.0

    for X_batch, Y_batch in train_loader:
        X_batch = X_batch.to(device)
        Y_batch = Y_batch.to(device)

        optimizer.zero_grad()

        predictions = model(X_batch)

        loss = criterion(
            predictions,
            Y_batch
        )

        loss.backward()
        optimizer.step()

        train_loss += loss.item()

    train_loss /= len(train_loader)

    # ---------- Validation ----------
    model.eval()
    val_loss = 0.0

    with torch.no_grad():
        for X_batch, Y_batch in val_loader:
            X_batch = X_batch.to(device)
            Y_batch = Y_batch.to(device)

            predictions = model(X_batch)

            loss = criterion(
                predictions,
                Y_batch
            )

            val_loss += loss.item()

    val_loss /= len(val_loader)

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    # Save best model
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = model.state_dict().copy()

    if (epoch + 1) % 10 == 0:
        print(
            f"Epoch {epoch+1:3d}/{epochs} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f}"
        )

# Restore best validation model
model.load_state_dict(best_state)

print("\nTraining complete!")
print("Best validation loss:", best_val_loss)
print("Device:", device)
```

```python
class OceanCNN_Hybrid(nn.Module):
    def __init__(self):
        super().__init__()

        self.cnn = nn.Sequential(
            nn.Conv2d(2, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),

            nn.MaxPool2d(2),

            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )

        self.spatial_fc = nn.Sequential(
            nn.Linear(32 * 4 * 4, 64),
            nn.ReLU(),
            nn.Dropout(0.3)
        )

        self.context_fc = nn.Sequential(
            nn.Linear(4, 16),
            nn.ReLU()
        )

        self.output = nn.Sequential(
            nn.Linear(64 + 16, 32),
            nn.ReLU(),
            nn.Linear(32, 6)
        )

    def forward(self, x, context):
        spatial_features = self.cnn(x)
        spatial_features = spatial_features.view(
            spatial_features.size(0), -1
        )

        spatial_features = self.spatial_fc(
            spatial_features
        )

        context_features = self.context_fc(
            context
        )

        combined = torch.cat(
            [spatial_features, context_features],
            dim=1
        )

        return self.output(combined)


model_hybrid = OceanCNN_Hybrid().to(device)

print(model_hybrid)
```

```python
sst_train_2000 = X_train_2000[:, 0, :, :]
sss_train_2000 = X_train_2000[:, 1, :, :]

sst_mean_2000 = np.nanmean(sst_train_2000)
sst_std_2000 = np.nanstd(sst_train_2000)

sss_mean_2000 = np.nanmean(sss_train_2000)
sss_std_2000 = np.nanstd(sss_train_2000)

X_train_2000_scaled = X_train_2000.copy()
X_val_2000_scaled = X_val_2000.copy()
X_test_2000_scaled = X_test_2000.copy()

X_train_2000_scaled[:, 0] = (X_train_2000_scaled[:, 0] - sst_mean_2000) / sst_std_2000
X_val_2000_scaled[:, 0] = (X_val_2000_scaled[:, 0] - sst_mean_2000) / sst_std_2000
X_test_2000_scaled[:, 0] = (X_test_2000_scaled[:, 0] - sst_mean_2000) / sst_std_2000

X_train_2000_scaled[:, 1] = (X_train_2000_scaled[:, 1] - sss_mean_2000) / sss_std_2000
X_val_2000_scaled[:, 1] = (X_val_2000_scaled[:, 1] - sss_mean_2000) / sss_std_2000
X_test_2000_scaled[:, 1] = (X_test_2000_scaled[:, 1] - sss_mean_2000) / sss_std_2000

target_2000_mean = Y_train_2000.mean()
target_2000_std = Y_train_2000.std()

Y_train_2000_scaled = (Y_train_2000 - target_2000_mean) / target_2000_std
Y_val_2000_scaled = (Y_val_2000 - target_2000_mean) / target_2000_std
Y_test_2000_scaled = (Y_test_2000 - target_2000_mean) / target_2000_std

print("SST mean/std:", sst_mean_2000, sst_std_2000)
print("SSS mean/std:", sss_mean_2000, sss_std_2000)
print("T_2000 mean/std:", target_2000_mean, target_2000_std)
print("Scaled shapes:", X_train_2000_scaled.shape, Y_train_2000_scaled.shape)
```

```python
class OceanCNN_Hybrid_2000(nn.Module):
    def __init__(self):
        super().__init__()

        self.cnn = nn.Sequential(
            nn.Conv2d(2,16,3,padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),

            nn.Conv2d(16,32,3,padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),

            nn.MaxPool2d(2),

            nn.Conv2d(32,32,3,padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )

        self.spatial_fc = nn.Sequential(
            nn.Linear(32*4*4,64),
            nn.ReLU(),
            nn.Dropout(0.3)
        )

        self.context_fc = nn.Sequential(
            nn.Linear(4,16),
            nn.ReLU()
        )

        self.output = nn.Sequential(
            nn.Linear(64+16,32),
            nn.ReLU(),
            nn.Linear(32,1)
        )

    def forward(self,x,context):
        spatial_features = self.cnn(x)
        spatial_features = spatial_features.view(spatial_features.size(0),-1)
        spatial_features = self.spatial_fc(spatial_features)

        context_features = self.context_fc(context)

        combined = torch.cat([spatial_features,context_features],dim=1)

        return self.output(combined)

model_2000 = OceanCNN_Hybrid_2000().to(device)

print(model_2000)
```

```python
criterion_2000 = nn.MSELoss()

optimizer_2000 = torch.optim.Adam(
    model_2000.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

best_val_loss_2000 = float("inf")
best_model_state_2000 = None
patience_2000 = 10
patience_counter_2000 = 0
num_epochs_2000 = 100

for epoch in range(num_epochs_2000):
    model_2000.train()
    train_loss_2000 = 0.0

    for X_batch, context_batch, Y_batch in train_loader_2000:
        X_batch = X_batch.to(device)
        context_batch = context_batch.to(device)
        Y_batch = Y_batch.to(device)

        optimizer_2000.zero_grad()

        predictions = model_2000(X_batch, context_batch)
        loss = criterion_2000(predictions, Y_batch)

        loss.backward()
        optimizer_2000.step()

        train_loss_2000 += loss.item()

    train_loss_2000 /= len(train_loader_2000)

    model_2000.eval()
    val_loss_2000 = 0.0

    with torch.no_grad():
        for X_batch, context_batch, Y_batch in val_loader_2000:
            X_batch = X_batch.to(device)
            context_batch = context_batch.to(device)
            Y_batch = Y_batch.to(device)

            predictions = model_2000(X_batch, context_batch)
            loss = criterion_2000(predictions, Y_batch)

            val_loss_2000 += loss.item()

    val_loss_2000 /= len(val_loader_2000)

    if val_loss_2000 < best_val_loss_2000:
        best_val_loss_2000 = val_loss_2000
        best_model_state_2000 = model_2000.state_dict()
        patience_counter_2000 = 0
    else:
        patience_counter_2000 += 1

    if (epoch + 1) % 10 == 0:
        print(
            f"Epoch {epoch+1}: "
            f"Train Loss = {train_loss_2000:.4f}, "
            f"Val Loss = {val_loss_2000:.4f}"
        )

    if patience_counter_2000 >= patience_2000:
        print(f"Early stopping at epoch {epoch+1}")
        break

model_2000.load_state_dict(best_model_state_2000)

print("Best validation loss:", best_val_loss_2000)
```

```python
def predict_subsurface_temperature(
    sst_patch,
    sss_patch,
    latitude,
    longitude,
    date
):
    model_hybrid.eval()

    sst_patch = np.asarray(sst_patch, dtype=np.float32)
    sss_patch = np.asarray(sss_patch, dtype=np.float32)

    satellite_patch = np.stack([sst_patch, sss_patch], axis=0)

    satellite_patch[0] = (
        satellite_patch[0] - 24.995323
    ) / 4.640743

    satellite_patch[1] = (
        satellite_patch[1] - 35.017105
    ) / 1.2542572

    date = pd.to_datetime(date)

    day_of_year = date.dayofyear

    day_sin = np.sin(
        2 * np.pi * day_of_year / 365.25
    )

    day_cos = np.cos(
        2 * np.pi * day_of_year / 365.25
    )

    context = np.array([
        latitude,
        longitude,
        day_sin,
        day_cos
    ], dtype=np.float32)

    context_mean = np.array([
        -14.9396105,
        73.7706833,
        0.00317567913,
        -0.00865530688
    ], dtype=np.float32)

    context_std = np.array([
        17.667383,
        20.706112,
        0.7036566,
        0.7104784
    ], dtype=np.float32)

    context = (
        (context - context_mean) / context_std
    )

    satellite_tensor = torch.tensor(
        satellite_patch,
        dtype=torch.float32
    ).unsqueeze(0).to(device)

    context_tensor = torch.tensor(
        context,
        dtype=torch.float32
    ).unsqueeze(0).to(device)

    with torch.no_grad():
        prediction = model_hybrid(
            satellite_tensor,
            context_tensor
        ).cpu().numpy()[0]

    prediction = (
        prediction * target_std + target_mean
    )

    return {
        depth: float(temp)
        for depth, temp in zip(
            [50, 100, 200, 500, 1000, 1500],
            prediction
        )
    }

print("Inference function created successfully.")
```

```python
def predict_2000_temperature(
    sst_patch,
    sss_patch,
    latitude,
    longitude,
    date
):
    model_2000.eval()

    sst_patch = np.asarray(sst_patch, dtype=np.float32)
    sss_patch = np.asarray(sss_patch, dtype=np.float32)

    satellite_patch = np.stack(
        [sst_patch, sss_patch],
        axis=0
    )

    satellite_patch[0] = (
        satellite_patch[0] - sst_mean_2000
    ) / sst_std_2000

    satellite_patch[1] = (
        satellite_patch[1] - sss_mean_2000
    ) / sss_std_2000

    date = pd.to_datetime(date)

    day_of_year = date.dayofyear

    day_sin = np.sin(
        2 * np.pi * day_of_year / 365.25
    )

    day_cos = np.cos(
        2 * np.pi * day_of_year / 365.25
    )

    context = np.array([
        latitude,
        longitude,
        day_sin,
        day_cos
    ], dtype=np.float32)

    context_mean_2000 = X_context_train_2000.mean(axis=0)
    context_std_2000 = X_context_train_2000.std(axis=0)

    context = (
        (context - context_mean_2000)
        / context_std_2000
    )

    satellite_tensor = torch.tensor(
        satellite_patch,
        dtype=torch.float32
    ).unsqueeze(0).to(device)

    context_tensor = torch.tensor(
        context,
        dtype=torch.float32
    ).unsqueeze(0).to(device)

    with torch.no_grad():
        prediction = model_2000(
            satellite_tensor,
            context_tensor
        ).cpu().numpy()[0, 0]

    prediction = (
        prediction * target_2000_std
        + target_2000_mean
    )

    return float(prediction)

print("2000m inference function created successfully.")
```
